```python
import heapq

def dijkstra(graph, source):
    n = len(graph)

    # dist[i] = shortest distance to vertex i
    dist = [float('inf')] * n

    # hops[i] = number of edges in the shortest path
    hops = [float('inf')] * n

    # parent[i] = previous vertex in the path
    parent = [-1] * n

    dist[source] = 0
    hops[source] = 0

    # (distance, hops, vertex)
    pq = [(0, 0, source)]

    while pq:
        current_dist, current_hops, u = heapq.heappop(pq)

        # Ignore outdated entries
        if (current_dist, current_hops) != (dist[u], hops[u]):
            continue

        for v, weight in graph[u]:

            new_dist = current_dist + weight
            new_hops = current_hops + 1

            # Prefer:
            # 1. Smaller distance
            # 2. If distance is same, fewer hops
            if (new_dist < dist[v] or
                (new_dist == dist[v] and new_hops < hops[v])):

                dist[v] = new_dist
                hops[v] = new_hops
                parent[v] = u

                heapq.heappush(
                    pq, (new_dist, new_hops, v)
                )

    return dist, hops, parent


def get_path(parent, source, destination):
    path = []
    current = destination

    while current != -1:
        path.append(current)
        current = parent[current]

    path.reverse()

    return path


# Graph with 6 vertices: 0 to 5
# (destination, weight)

graph = [
    [(1, 2), (2, 4)],       # 0
    [(2, 2), (3, 5)],       # 1
    [(3, 2), (4, 4)],       # 2
    [(4, 1), (5, 3)],       # 3
    [(5, 2)],               # 4
    []                      # 5
]

source = 0

dist, hops, parent = dijkstra(graph, source)

print("Shortest paths from vertex", source)
print("-" * 45)

for v in range(len(graph)):
    if dist[v] == float('inf'):
        print(f"Vertex {v}: No path")
    else:
        path = get_path(parent, source, v)

        print(
            f"Vertex {v}: "
            f"Distance = {dist[v]}, "
            f"Hops = {hops[v]}, "
            f"Path = {' -> '.join(map(str, path))}"
        )
```
