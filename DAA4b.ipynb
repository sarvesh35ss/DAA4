```python
import heapq

def dijkstra(graph, source, destination):
    n = len(graph)

    # Shortest distance to each node
    dist = [float('inf')] * n

    # To reconstruct the shortest path
    parent = [-1] * n

    dist[source] = 0

    # Priority queue: (distance, node)
    pq = [(0, source)]

    while pq:
        current_dist, u = heapq.heappop(pq)

        if current_dist > dist[u]:
            continue

        for v, weight in graph[u]:
            new_dist = current_dist + weight

            if new_dist < dist[v]:
                dist[v] = new_dist
                parent[v] = u
                heapq.heappush(pq, (new_dist, v))

    # Construct path
    path = []
    current = destination

    while current != -1:
        path.append(current)
        current = parent[current]

    path.reverse()

    return dist[destination], path


# City road network
# (destination, travel time in minutes)

graph = [
    [(1, 10), (2, 5)],       # Node 0 - Airport
    [(2, 2), (3, 1)],        # Node 1
    [(1, 3), (3, 9), (4, 2)],# Node 2
    [(4, 4)],                # Node 3
    []                       # Node 4 - Central Station
]

source = 0
destination = 4

distance, path = dijkstra(graph, source, destination)

print("Fastest travel time:", distance, "minutes")
print("Fastest route:", " -> ".join(map(str, path)))
```
