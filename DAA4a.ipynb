```python
import heapq

def dijkstra(graph, source):
    n = len(graph)
    dist = [float('inf')] * n
    parent = [-1] * n

    dist[source] = 0
    pq = [(0, source)]  # (distance, vertex)

    while pq:
        d, u = heapq.heappop(pq)

        if d > dist[u]:
            continue

        for v, weight in graph[u]:
            new_dist = d + weight

            if new_dist < dist[v]:
                dist[v] = new_dist
                parent[v] = u
                heapq.heappush(pq, (new_dist, v))

    return dist, parent


def get_path(parent, source, destination):
    path = []
    current = destination

    while current != -1:
        path.append(current)
        if current == source:
            break
        current = parent[current]

    if path[-1] != source:
        return []

    return path[::-1]


# Directed weighted graph with vertices 0 to 5
graph = [
    [(1, 4), (2, 2)],       # 0
    [(2, 1), (3, 5)],       # 1
    [(1, 1), (3, 8), (4, 10)], # 2
    [(4, 2), (5, 6)],       # 3
    [(5, 3)],               # 4
    []                      # 5
]

source = 0

dist, parent = dijkstra(graph, source)

print("Shortest paths from vertex 0:\n")

for vertex in range(len(graph)):
    if dist[vertex] == float('inf'):
        print(f"0 -> {vertex}: No path")
    else:
        path = get_path(parent, source, vertex)
        print(f"0 -> {vertex}: Distance = {dist[vertex]}, Path = {' -> '.join(map(str, path))}")
```
